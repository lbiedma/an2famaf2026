ITEM (a) Info importante:
* El núcleo de $A^TA$ se define como $Ker(A^TA) = \{ x \in \mathbb{R}^n/ A^TAx =0\}$
* El núcleo de $A$ se define como $Ker(A) = \{ x \in \mathbb{R}^n/ Ax =0\}$


(1) Vamos a probar primero $Ker(A)\subseteq Ker(A^TA)$

Sea $x \in Ker(A)$ entonces,

$$
Ax = 0 \Rightarrow A^TAx = A^T0 \Rightarrow A^TAx = 0 \Rightarrow  x \in Ker(A^TA)
$$
Listo!. Falta el otro lado.

(2) Vamos a probar ahora $Ker(A^TA)\subseteq Ker(A)$

Sea $x \in Ker(A^TA)$ entonces,

$$
A^TAx = 0 \overset{(1)}{\Rightarrow} x^TA^TAx = x^T0 = 0 \overset{(2)}{\Rightarrow} || Ax ||_2 = 0\; \text{o}\; < Ax,Ax > = 0 \Rightarrow Ax = 0\Rightarrow  x \in Ker(A)
$$

$(1)$ multiplicar por $x^T$ en ambos miembros.

$(2)$ definición de norma o de punto interior (cualquiera de las dos vale!)

Luego, $Ker(A^TA) = Ker(A^T)$
