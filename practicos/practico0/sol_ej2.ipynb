ITEM (a)

Info importante:
Cada columna de $AB$ es de la forma $Ab_j$, donde $b_j$ es la $j$-ésima columna de $B$. Si $a_1,\dots,a_r$ son las columnas de $A$, entonces
$$
Ab_j = \sum_{i=1}^r (b_j)_i \, a_i,
$$
por lo que cada columna de $AB$ es combinación lineal de las columnas de $A$. Luego,
$$
\text{Im}(AB) \subseteq \text{Im}(A) \Rightarrow rg(AB) = dim (\text{Im}(AB)) \leq dim (\text{Im}(A)) = rg(A).
$$

Por otra parte, si $a_i^T$ denota la $i$-ésima fila de $A$, la $i$-ésima fila de $AB$ es $a_i^T B$, combinación lineal de las filas de $B$. Entonces
$$
\text{Im}((AB)^T) \subseteq \text{Im}(B^T) \implies rg(AB) \leq rg(B),
$$
Listo!


ITEM (b) 

Info importante: Cada columna de $A+B$ es suma de la columna correspondiente de $A$ más la de $B$, por lo que pertenece a $\text{Im}(A) + \text{Im}(B)$. Así,
$$
\text{Im}(A+B) \subseteq \text{Im}(A) + \text{Im}(B).
$$
Luego, por propiedades de dim,
$$
rg(A+B) \leq \text{dim}\big(\text{Im}(A)+\text{Im}(B)\big) \leq \text{dim}(\text{Im}(A)) + \text{dim}(\text{Im}(B)) = rg(A) + rg(B).
$$
Listo!

ITEM (c)

Info importante: Vamos a usar ahora $Ker(A^TA) = Ker(A)$.

Como $A$ y $A^TA$ tienen ambas $n$ columnas entonces, 

$$
rg(A^TA) = n - \text{dim}(Ker(A^TA)) = n - \text{dim}(Ker(A)) = rg(A).
$$

Aplicando lo mismo para $A^T$ se llega a la siguiente igualdad
$$
rg(AA^T) = rg(A^T).
$$

Para probar $rg(A) = rg(A^T)$ vamos a usar (a):

* $rg(A^TA) \leq rg(A^T)$ y como $rg(A^TA)=rg(A)$, entonces $rg(A) \leq rg(A^T)$,
* $rg(AA^T) \leq rg(A)$ y como $rg(AA^T)=rg(A^T)$ entonces $rg(A^T) \leq rg(A)$.

Luego, $rg(A) = rg(A^T)$.

Listo!


ITEM (d)
Info importante:
* Como $rg(A) = r$ y $rg(B) = r$, entonces por (c) $rg(A^TA) = rg(BB^T)$. Como son matrices $r\times r$, son invertibles, al igual que el producto $A^TABB^T$ Entonces,
$$
r = rg(A^TABB^T) \leq rg(AB)
$$
esto se da por (a). Por otra parte, sabemos que,
$$
rg(AB) \leq rg(A) = r,
$$
Luego, $rg(AB) =r$. Listo!
