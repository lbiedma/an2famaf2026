## Caso $n=1$:


Sea $ A = 1$ entonces $A^{-1} = 1$, $A^{-1}A = 1$, luego existe $A^{-1}$ triangular superior con unos en la diagonal.


## Hipótesis inductiva

Suponemos que por hipótesis inductiva, $A_{n-1,n-1}\in\R^{(n-1)\times(n-1)}$ triangular superior con unos en la diagonal, existe $A_{n-1,n-1}^{-1}\in\mathbb{R}^{(n-1)\times(n-1)}$ triangular superior con unos en la diagonal.

## Caso $n$

Sea ahora $A\in\mathbb{R}^{n\times n}$ invertible. Podemos escribir

$$
A=
\begin{pmatrix}
A_{n-1,n-1} & a^{n-1}\\
0_{n-1}^{T} & 1
\end{pmatrix},
$$

donde $a^{n-1}\in\mathbb{R}^{n-1},$ $0_{n-1} \in \R^{n-1}$ un vector de ceros. Construimos:
$$
A^{-1}
=
\begin{pmatrix}
A_{n-1,n-1}^{-1}
&
-A_{n-1,n-1}^{-1}a^{n-1}
\\[8pt]
0_{n-1}^{T}
& 1
\end{pmatrix}.
$$



Verificamos que esta matriz es realmente la inversa:

$$
AA^{-1}
=
\begin{pmatrix}
A_{n-1,n-1} & a^{n-1}\\
0_{n-1}^{T} & 1
\end{pmatrix}
\begin{pmatrix}
A_{n-1,n-1}^{-1}
&
-A_{n-1,n-1}^{-1}a^{n-1}
\\[8pt]
0_{n-1}^{T}
&1
\end{pmatrix}.
$$

Multiplicando por bloques,

$$
AA^{-1}
=
\begin{pmatrix}
I_{n-1} & 0_{n-1}\\
0_{n-1}^{T} & 1
\end{pmatrix}
=I_n.
$$

De forma análoga, $A^{-1}A=I_n$. Luego,

$$
A^{-1}=
\begin{pmatrix}
A_{n-1,n-1}^{-1}
&
-A_{n-1,n-1}^{-1}a^{n-1}
\\[8pt]
0_{n-1}^{T}
&
1
\end{pmatrix}
$$

es triangular superior con unos en la diagonal.
