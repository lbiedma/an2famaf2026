# Item b)

## Caso $n = 3$

$$
M_1^{-1} M_2^{-1} = (I+v^1(e^1)^T)(I+v^2(e^2)^T) = I+v^1(e^1)^T+v^2(e^2)^T+v^1(e^2)^Tv^1(e^2)^T  
$$

donde $v^1(e^1)^Tv^2(e^2)^T = 0$ (matriz de ceros) pues los productos entre los elemenstos se cruzan con ceros. Luego, 

$$
M_1^{-1} M_2^{-1} = I+v^1(e^1)^T+v^2(e^2)^T 
$$

## Hipótesis inductiva

$$
M_1^{-1} \ldots M_{n-2}^{-1} = \left(I+\underset{k=1}{\overset{n-2}{\sum}}v^k(e^k)^T\right)
$$

## Caso $n$

$$
M_1^{-1} \ldots M_{n-2}^{-1} M_{n-1}^{-1} = \left(I+\underset{k=1}{\overset{n-2}{\sum}}v^k(e^k)^T\right) M_{n-1}^{-1} = \left(I+\underset{k=1}{\overset{n-2}{\sum}}v^k(e^k)^T\right)(I+v^{n-1}(e^{n-1})^T)
$$

$$
M_1^{-1} \ldots M_{n-2}^{-1} M_{n-1}^{-1} = I + v^{n-1}(e^{n-1})^T + \underset{k=1}{\overset{n-2}{\sum}}v^k(e^k)^T +\left(\underset{k=1}{\overset{n-2}{\sum}}v^k(e^k)^T\right)(v^{n-1}(e^{n-1})^T)
$$

El termino $\left(\underset{k=1}{\overset{n-2}{\sum}}v^k(e^k)^T\right)(v^{n-1}(e^{n-1})^T) = 0$ (matriz de ceros) por la misma razón del caso base. Los elementos de las matrices se cruzan con los cero anulando la matriz resultantes.

Luego, 

$$
M_1^{-1} \ldots M_{n-2}^{-1} M_{n-1}^{-1} = I + \underset{k=1}{\overset{n-1}{\sum}}v^k(e^k)^T
$$
