Sean los conjuntos de indices $\mathcal{I} = \{1,2, \ldots, k\}$ y $\mathcal{J} = \{k+1,k+2, \ldots, n\}$ 

Definimos: 

$$
H = 
\begin{bmatrix}
I_{\mathcal{I}\mathcal{I}}\\
0_{\mathcal{J}\mathcal{I}}
\end{bmatrix}
$$

donde $I_{\mathcal{I}\mathcal{I}} \in \R^{k\times k}$, es una matriz identidad y $0_{\mathcal{J}\mathcal{I}} \in \R^{(n-k)\times k}$ una matriz de ceros.  

La matriz $B$ se puede indexar de la siguiente forma: 
$$
B = 
\begin{bmatrix}
B_{\mathcal{I}\mathcal{I}} & B_{\mathcal{I}\mathcal{J}}\\
B_{\mathcal{J}\mathcal{I}} & B_{\mathcal{J}\mathcal{J}} \\
\end{bmatrix}
$$

Hacemos la cuenta $H^TBH$:

$$ 
H^{T}BH = 
\begin{bmatrix}
I_{\mathcal{I}\mathcal{I}} & 0_{\mathcal{I}\mathcal{J}} \\
\end{bmatrix}

\begin{bmatrix}
B_{\mathcal{I}\mathcal{I}} & B_{\mathcal{I}\mathcal{J}}\\
B_{\mathcal{J}\mathcal{I}} & B_{\mathcal{J}\mathcal{J}} \\
\end{bmatrix} 

\begin{bmatrix}
I_{\mathcal{I}\mathcal{I}}\\
0_{\mathcal{J}\mathcal{I}}
\end{bmatrix}
= 

\begin{bmatrix}
B_{\mathcal{I}\mathcal{I}} & B_{\mathcal{I}\mathcal{J}}\\
\end{bmatrix} 
\begin{bmatrix}
I_{\mathcal{I}\mathcal{I}}\\
0_{\mathcal{J}\mathcal{I}}
\end{bmatrix}


= B_{\mathcal{I}\mathcal{I}}
$$


Entonces si reemplazamos $B$ por $LA$ tenemos que:

$$
B_{\mathcal{I}\mathcal{I}} =
H^{T}BH =
H^{T}LAH =   
\begin{bmatrix}
I_{\mathcal{I}\mathcal{I}} & 0_{\mathcal{I}\mathcal{J}} \\
\end{bmatrix}

\begin{bmatrix}
L_{\mathcal{I}\mathcal{I}} & O_{\mathcal{I}\mathcal{J}}\\
L_{\mathcal{J}\mathcal{I}} & L_{\mathcal{J}\mathcal{J}} \\
\end{bmatrix}

\begin{bmatrix}
A_{\mathcal{I}\mathcal{I}} & A_{\mathcal{I}\mathcal{J}}\\
A_{\mathcal{J}\mathcal{I}} & A_{\mathcal{J}\mathcal{J}} \\
\end{bmatrix} 

\begin{bmatrix}
I_{\mathcal{I}\mathcal{I}}\\
0_{\mathcal{J}\mathcal{I}}
\end{bmatrix}
$$


$$
B_{\mathcal{I}\mathcal{I}} =
\begin{bmatrix}
L_{\mathcal{I}\mathcal{I}} & O_{\mathcal{I}\mathcal{J}}\\
\end{bmatrix}

\begin{bmatrix}
A_{\mathcal{I}\mathcal{I}} \\
A_{\mathcal{J}\mathcal{I}} \\
\end{bmatrix} 
= L_{\mathcal{I}\mathcal{I}}A_{\mathcal{I}\mathcal{I}}
$$


Por lo tanto,
$$
B_{\mathcal{I}\mathcal{I}} = L_{\mathcal{I}\mathcal{I}}A_{\mathcal{I}\mathcal{I}}
$$
