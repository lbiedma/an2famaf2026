Sean $A = U\Sigma V^T$ y $A^T = V \Sigma^T U^T$  

$\Sigma = 
\begin{bmatrix}
\Sigma_n \\
0 \\
\end{bmatrix}
\in \R^{m\times n}$
 con 
$\Sigma_n = \begin{bmatrix}
\sigma_1 & &\\
& \sigma_2 &\\
& & \ddots & \\
& &  & \sigma_n\\
\end{bmatrix}$

$|| (A^TA)^{-1} ||_2 = || (V\Sigma^T U^T U \Sigma V^T)^{-1}||_2 = 
|| 
(V 
\begin{bmatrix}
\Sigma_n^T & 0^T 
\end{bmatrix}
\begin{bmatrix}
\Sigma_n \\
 0 
\end{bmatrix}
 V^T)^{-1}||_2
 = ||(V \Sigma_n^T \Sigma_n V^T)^{-1}||_2 
 = ||\Sigma_n^{-1} (\Sigma_n^T)^{-1}||_2 = \sigma_n^{-2}$

 

como $\sigma_n$ es el más chico entonces $\sigma_n^{-2}$ pasa a ser el valor más grande entre los valores singulares. 

Puebo ahora, $||A(A^TA)^{-1}A^T ||_2 = 1$



$||A(A^TA)^{-1}A^T ||_2 = ||U \Sigma V^T (V \Sigma_n^T \Sigma_n V^T)^{-1} V \Sigma^T U^T||_2
= ||U\Sigma \Sigma_n^{-1} (\Sigma_n^T)^{-1} \Sigma^T U^T||_2 = 
||U 
 \begin{bmatrix}
I_n & 0 \\
0 & 0 \\
\end{bmatrix}
 U^T||_2 = 1$
