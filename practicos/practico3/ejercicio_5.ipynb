veamos primero:
$$
|| A||_2 = \underset{||x||_2 =1}{max}||Ax||_2
$$

La definición de norma inducida: 
$$
||A||_2 = \underset{x\neq 0}{sup} \dfrac{||Ax||_2}{||x||_2} 
$$

Sea $\hat{x}$ y definimos $x = \dfrac{\hat{x}}{|| \hat{x}||_2}$, entonces: 

$$
||A||_2 = \underset{x\neq 0}{sup} \dfrac{||A\hat{x}||_2}{||\hat{x}||_2} =  
\underset{x\neq 0}{sup}||A\dfrac{\hat{x}}{||\hat{x}||_2}||_2 = \underset{||x||_2 =1}{max}||Ax||_2
$$

Veamos ahora, 

$$
\underset{||y||_2 =||x||_2 =1}{max}||Ax||_2 = \underset{||y||_2 =||x||_2 =1}{max}\; y^TAx
$$

tomemos $y= \dfrac{Ax}{||Ax||_2}$ tal que $||y||_2 = 1$, entonces

$$
\underset{||y||_2 =||x||_2 =1}{max}\; y^TAx =\underset{||y||_2 =||x||_2 =1}{max}\; \dfrac{(Ax)^T}{||Ax||_2}Ax = \underset{||y||_2 =||x||_2 =1}{max}\; \dfrac{x^TA^TAx}{||Ax||_2} =  \underset{||y||_2 =||x||_2 =1}{max}\; \dfrac{||Ax||_2^2}{||Ax||_2} \underset{||y||_2 =||x||_2 =1}{max}||Ax||_2 
$$

donde $||Ax||_2^2 = x^TA^TAx$
