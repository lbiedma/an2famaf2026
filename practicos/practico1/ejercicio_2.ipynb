## a) $\Rightarrow$ b)

$A$ es simétrica y definida positiva, es decir,

$$
A=A^T,
\qquad
x^T A x>0,\quad x\neq 0.
$$

Sea $v\in\mathbb{R}^n$, $v\neq 0$, tal que

$$
Av=\lambda v,
\qquad \lambda\in\mathbb{R}.
$$

Entonces,

$$
v^TAv=v^T(\lambda v)
=\lambda v^Tv
=\lambda\|v\|^2.
$$

Como $A$ es definida positiva,

$$
v^TAv>0,
$$

y como $v\neq0$,

$$
\|v\|^2>0.
$$

Por lo tanto,

$$
\lambda>0.
$$

## b) $\Rightarrow$ c)

$A$ es simétrica y los autovalores de $A$ son positivos.

Debo probar que las submatrices $A_{II}$ son definidas positivas, con

$$
I=\{1,2,\ldots,n\}.
$$

Como $A\in\mathbb{R}^{n\times n}$ es simétrica, si y solo si $A$ es diagonalizable. Entonces existen una matriz $P$ ortogonal y una matriz $D$ diagonal tales que

$$
P^TAP=D.
$$

Donde los autovalores de $A$ están en la diagonal de $D$.

Sea

$$
x\neq0,\qquad x\in\mathbb{R}^m.
$$

Consideramos

$$
x^TA_{II}x
$$

y, usando la matriz de inclusión $\widetilde I$,

$$
x^T A_{II}x
=x^T\widetilde I^T A\widetilde I x.
$$

Como $A=PDP^T$,

$$
x^T A_{II}x
=x^T\widetilde I^T P D P^T\widetilde I x.
$$

Definimos

$$
y=P^T\widetilde I x.
$$

Entonces,

$$
x^T A_{II}x=y^TDy.
$$

Como $D$ es diagonal,

$$
y^TDy
=\sum_{i=1}^{n}\lambda_i y_i^2.
$$

Como todos los autovalores son positivos,

$$
\lambda_i>0,
$$

y, para $y\neq0$,

$$
y^TDy
=\sum_{i=1}^{n}\lambda_i y_i^2>0.
$$

Por lo tanto,

$$
y^TDy>0
\quad\Longrightarrow\quad
x^TA_{II}x>0.
$$

Por tanto, $A_{II}$ es definida positiva.

## c) $\Rightarrow$ a)

Trivial: $A$ es simétrica.
