## Primera forma

Sea

$$
x=
\begin{pmatrix}
x_1\\
x_2
\end{pmatrix}
\in\mathbb{R}^n,
\qquad
x_1\in\mathbb{R}^{n_1},\quad
x_2\in\mathbb{R}^{n_2},
$$

con $x\neq0$.

Como $A$ es definida positiva,

$$
x^TAx>0.
$$

Desarrollando por bloques,

$$
\begin{pmatrix}x_1^T & x_2^T\end{pmatrix}
\begin{pmatrix}
A_{11} & A_{12}\\
A_{21} & A_{22}
\end{pmatrix}
\begin{pmatrix}x_1\\x_2\end{pmatrix}>0.
$$

Entonces,

$$
\begin{pmatrix}
x_1^TA_{11}+x_2^TA_{21} &
x_1^TA_{12}+x_2^TA_{22}
\end{pmatrix}
\begin{pmatrix}x_1\\x_2\end{pmatrix}>0,
$$

y por lo tanto

$$
x_1^TA_{11}x_1+x_2^TA_{21}x_1
+x_1^TA_{12}x_2+x_2^TA_{22}x_2>0.
$$

Definimos

$$
C=
\begin{pmatrix}
0 & -A_{11}^{-1}A_{12}\\
0 & I
\end{pmatrix}.
$$

Tomemos $Cx=x$ para la elección de $x$ que nos interesa. Es decir,

$$
\begin{pmatrix}x_1\\x_2\end{pmatrix}
=
\begin{pmatrix}
0 & -A_{11}^{-1}A_{12}\\
0 & I
\end{pmatrix}
\begin{pmatrix}x_1\\x_2\end{pmatrix}
=
\begin{pmatrix}
-A_{11}^{-1}A_{12}x_2\\
x_2
\end{pmatrix}.
$$

Luego,

$$
x_1=-A_{11}^{-1}A_{12}x_2.
$$

Reemplazando en la expresión de $x^TAx$:

$$
\begin{aligned}
x^TAx={}&
(-A_{11}^{-1}A_{12}x_2)^T A_{11}
(-A_{11}^{-1}A_{12}x_2)\\
&+x_2^TA_{21}(-A_{11}^{-1}A_{12}x_2)\\
&+(-A_{11}^{-1}A_{12}x_2)^TA_{12}x_2\\
&+x_2^TA_{22}x_2.
\end{aligned}
$$

Como $A$ es simétrica,

$$
A_{12}^T=A_{21}.
$$

Entonces,

$$
\begin{aligned}
x^TAx={}&
x_2^TA_{12}^T A_{11}^{-1}A_{12}x_2
-x_2^TA_{21}A_{11}^{-1}A_{12}x_2\\
&-x_2^TA_{12}^T A_{11}^{-1}A_{12}x_2
+x_2^TA_{22}x_2.
\end{aligned}
$$

Los términos se simplifican y queda

$$
x^TAx
=-x_2^TA_{21}A_{11}^{-1}A_{12}x_2+x_2^TA_{22}x_2.
$$

Por lo tanto,

$$
x^TAx
=x_2^T\left(A_{22}-A_{21}A_{11}^{-1}A_{12}\right)x_2
=x_2^TSx_2>0.
$$

Así, $S$ es definida positiva.

## Segund forma

Vamos a construir una matriz $B$ tal que

$$
B^TAB=D,
$$

de modo que $D$ sea una matriz diagonal por bloques.

Tomamos

$$
B=
\begin{pmatrix}
B_{11} & B_{12}\\
B_{21} & B_{22}
\end{pmatrix}.
$$

Entonces,

$$
AB=
\begin{pmatrix}
A_{11}B_{11}+A_{12}B_{21} &
A_{11}B_{12}+A_{12}B_{22}\\
A_{21}B_{11}+A_{22}B_{21} &
A_{21}B_{12}+A_{22}B_{22}
\end{pmatrix}.
$$

Queremos que

$$
A_{11}B_{12}+A_{12}B_{22}=0.
$$

Tomamos $B_{22}=I$. Entonces,

$$
B_{12}=-A_{11}^{-1}A_{12}.
$$

Con esta elección,

$$
AB=
\begin{pmatrix}
A_{11}B_{11}+A_{12}B_{21} & 0\\
A_{21}B_{11}+A_{22}B_{21} &
-A_{21}A_{11}^{-1}A_{12}+A_{22}
\end{pmatrix}.
$$

Como

$$
S=A_{22}-A_{21}A_{11}^{-1}A_{12},
$$

queda

$$
AB=
\begin{pmatrix}
A_{11}B_{11}+A_{12}B_{21} & 0\\
A_{21}B_{11}+A_{22}B_{21} & S
\end{pmatrix}.
$$

Vamos a tomar $B_{21}=0$ para facilitar las cuentas. Entonces,

$$
B_{11}=I,
$$

y finalmente

$$
B=
\begin{pmatrix}
I & -A_{11}^{-1}A_{12}\\
0 & I
\end{pmatrix}.
$$

Calculamos ahora $B^TAB$:

$$
B^T=
\begin{pmatrix}
I & 0\\
-A_{21}A_{11}^{-1} & I
\end{pmatrix},
$$

donde usamos que $A$ es simétrica.

Por lo tanto,

$$
B^TAB=
\begin{pmatrix}
A_{11} & 0\\
0 & S
\end{pmatrix}
=D.
$$

Vamos a usar el ejercicio 3a. La matriz $B$ es triangular por bloques con identidad en la diagonal, por lo que

$$
\det(B)=1\neq0.
$$

Entonces $B$ tiene columnas linealmente independientes y, por tanto, rango completo.

Como

$$
D=B^TAB,
$$

se concluye que $D$ es definida positiva.

Luego, 

$$
S=A_{22}-A_{21}A_{11}^{-1}A_{12}
$$

es definida positiva.
